```python
# Databricks notebook source
from pyspark import SparkConf, SparkContext

# COMMAND ----------

conf = SparkConf().setAppName("Read File")

# COMMAND ----------

sc = SparkContext.getOrCreate(conf=conf)


```

```python
# COMMAND ----------

rdd = sc.textFile(r"..\..\SampleData\sample_words.txt")

# COMMAND ----------

print('\n\n\n')
print(rdd.collect())
print('\n\n\n')
```

```python
rdd2 = rdd.map(lambda x: x.split(" "))
print(rdd2.collect())
```

```python
rdd3 = rdd.flatMap(lambda x: x.split(" "))
print(rdd3.collect())
```

```python
rdd4 = rdd3.filter(lambda x: len(x) > 3)
print(rdd4.collect())
```

```python
rdd5 = rdd4.distinct()
print(rdd5.collect())
```

```python
dictv = rdd5.countByValue()
print(dictv)
```

```python
rdd5.getNumPartitions()
```

```python
rdd5.repartition(5)
rdd5.saveAsTextFile('/FileStore/tables/output/3partitionOutput')
```

```python
rdd = sc.textFile('movie_ratings.csv')
print(rdd.collect())
```

```python
rdd2 = rdd.map(lambda line:line.split(","))
rdd3 = rdd2.map(lambda tokens: (tokens[0], (float(tokens[1]),1)))
rdd4 = rdd3.reduceByKey(lambda a,b: (a[0]+b[0], a[1]+b[1]))
rdd5 = rdd4.map(lambda x: (x[0], x[1][0]/x[1][1]))
rdd5.collect()
```

```python
rdd2 = rdd.map(lambda x: (x.split(',')[0],   (int(x.split(',')[1]),1)  ))
rdd3 = rdd2.reduceByKey(lambda x,y: (x[0]+y[0], x[1]+y[1]))
rdd4 = rdd3.map(lambda x: (x[0],x[1][0]/x[1][1]))
rdd4.collect()
```

```python
# min and max
rdd2 = rdd.map(lambda line:line.split(","))
rdd3 = rdd2.map(lambda tokens: (tokens[0], (float(tokens[1]),float(tokens[1]))))
rdd4 = rdd3.reduceByKey(lambda a,b: (min(a[0],b[0]), max(a[1],b[1])))
rdd4.collect()
```

```python
# wrong way
rdd2 = rdd.map(lambda line:line.split(","))
rdd3 = rdd2.map(lambda tokens: (tokens[0], float(tokens[1])))
rdd4 = rdd3.reduceByKey(lambda a,b: (min(a,b), max(a,b)))
```
